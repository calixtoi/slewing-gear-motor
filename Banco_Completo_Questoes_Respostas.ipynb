## Questão 1.3: Equação Normal

**Nível:** Difícil  
**Tema:** Derivação da Equação Normal

Derive a Equação Normal começando de:
$$\mathcal{L}(\theta) = \|B\theta - y\|^2$$

Mostre TODOS os passos.

**Resposta:**

**Passo 1:** Expanda o quadrado
$$\mathcal{L}(\theta) = (B\theta - y)^T(B\theta - y)$$

**Passo 2:** Distribua
$$\mathcal{L}(\theta) = \theta^T B^T B\theta - 2\theta^T B^T y + y^T y$$

**Passo 3:** Calcule o gradiente
$$\frac{\partial \mathcal{L}}{\partial \theta} = 2B^T B\theta - 2B^T y$$

**Passo 4:** Iguale a zero (condição de mínimo)
$$2B^T B\theta - 2B^T y = 0$$
$$B^T B\theta = B^T y$$

**Passo 5:** Resolva para θ
$$\theta^* = (B^T B)^{-1}B^T y$$

## Questão 1.4: Funções de Base

**Nível:** Médio  
**Tema:** Tipos de Funções de Base

Para cada tipo de função de base, descreva:
- A fórmula matemática
- Um exemplo prático de quando usar
- Uma vantagem e uma desvantagem

a) Polinomial
b) RBF Gaussiana
c) Fourier

**Resposta:**

**a) Polinomial:**
- **Fórmula:** $B_j(x) = x^j$ para $j = 0, 1, 2, \ldots, d$
- **Quando usar:** Dados que parecem seguir padrões suaves (curvas polinomiais)
- **Vantagem:** Simples, interpretável, funciona bem para dados globalmente suaves
- **Desvantagem:** Alto grau causa instabilidade numérica (Runge phenomenon)

**b) RBF Gaussiana:**
- **Fórmula:** $B_j(x) = \exp\left(-\frac{(x-c_j)^2}{2\sigma^2}\right)$
- **Quando usar:** Dados com estrutura local (múltiplos "bumps" em lugares diferentes)
- **Vantagem:** Captura bem features localizadas, evita instabilidade polinomial
- **Desvantagem:** Precisa escolher centros $c_j$ e largura $\sigma$

**c) Fourier:**
- **Fórmula:** $B_j(x) = \sin(jx)$ ou $\cos(jx)$
- **Quando usar:** Dados periódicos ou com padrões oscilatórios (séries temporais, ondas)
- **Vantagem:** Excelente para dados periódicos, convergência rápida
- **Desvantagem:** Ruim para dados não-periódicos (não consegue capturar tendências globais)

## Questão 1.5: Singularidade de B^T B

**Nível:** Médio  
**Tema:** Matriz de Design Singular

Quando a matriz $B^T B$ é singular (não invertível)? Dê 2 exemplos práticos.

**Resposta:**

$B^T B$ é singular quando as **colunas de B são linearmente dependentes**.

**Exemplo 1: Menos dados que features**
- Temos n = 2 observações
- p = 3 features
- B é 2×4 (n < p+1)
- Impossível encontrar solução única para 4 incógnitas com apenas 2 equações
- $B^T B$ é 4×4 singular

**Exemplo 2: Colinearidade perfeit**
- Feature 1: [1, 2, 3]
- Feature 2: [2, 4, 6] (exatamente 2× Feature 1)
- As duas features carregam exatamente a mesma informação
- As colunas de B são linearmente dependentes

**Como Ridge Resolve:**
- Ridge adiciona $\lambda I$ à diagonal:
  $$(B^T B + \lambda I)^{-1}$$
- O termo $\lambda I$ garante que a matriz tem "alguns" valores próprios não-nulos
- Isto torna a matriz invertível, mesmo se $B^T B$ era singular

## Questão 2.2: MSE Decomposition

**Nível:** Médio  
**Tema:** Decomposição do Erro

Enuncie a decomposição do MSE em 3 componentes. Para cada um, diga:
- Como é gerado?
- Como aparece no trade-off viés-variância?
- Como reduzir?

**Resposta:**

$$\text{MSE} = \text{Viés}^2 + \text{Variância} + \text{Erro Irreduzível}$$

**1. VIÉS² (Bias²)**
- **Gerado por:** Modelo muito simples/restritivo
- **No trade-off:** Diminui quando aumentamos complexidade do modelo
- **Como reduzir:**
  - Usar modelo mais complexo
  - Mais features/funções de base
  - Polynomial degree maior

**2. VARIÂNCIA (Variance)**
- **Gerado por:** Modelo muito sensível aos dados de treino específicos
- **No trade-off:** Aumenta quando aumentamos complexidade do modelo
- **Como reduzir:**
  - Regularização (Ridge, Lasso)
  - Mais dados de treinamento
  - Modelo mais simples
  - Early stopping

**3. ERRO IRREDUZÍVEL (Irreducible Error)**
- **Gerado por:** Ruído aleatório nos dados (não controlável)
- **No trade-off:** Constante, não muda com modelo
- **Como reduzir:** 
  - Não pode reduzir!
  - Só esperamos atingir este limite
  - Indica quando parar de otimizar

## Questão 2.3: Trade-off Viés-Variância

**Nível:** Médio  
**Tema:** O Dilema Fundamental

Explique por que existe um trade-off entre viés e variância. O que acontece quando:
a) Aumentamos a complexidade do modelo?
b) Diminuímos a complexidade do modelo?

**Resposta:**

**Por que existe o trade-off:**

Um modelo complexo consegue se ajustar melhor aos dados (reduz viés), mas ao fazer isso, fica mais sensível a flutuações e ruído nos dados (aumenta variância). É impossível minimizar ambos simultaneamente - há um "sweet spot" entre as duas extremidades.

**a) Aumentar complexidade:**
- ✓ **Viés diminui:** O modelo mais flexível consegue capturar padrões mais complexos
- ✗ **Variância aumenta:** Também se ajusta ao ruído específico
- **Resultado:** Erro total pode aumentar se variância cresce mais que viés diminui

**b) Diminuir complexidade:**
- ✗ **Viés aumenta:** O modelo muito simples não consegue capturar padrões
- ✓ **Variância diminui:** Menos sensível ao ruído
- **Resultado:** Erro total aumenta se viés cresce demais

**Meta:** Encontrar o ponto ótimo onde (Viés² + Variância) é minimizado.

## Questão 3.2: Elastic Net

**Nível:** Fácil  
**Tema:** Combinação de Regularizadores

O que é Elastic Net? Como combina Ridge e Lasso?

**Resposta:**

**Elastic Net** é a combinação de ambos os regularizadores:

$$\mathcal{L}(\theta) = \frac{1}{N}\|B\theta-y\|^2 + \lambda_1\|\theta\|_1 + \lambda_2\|\theta\|^2$$

**Como combina:**
- **Termo L¹ ($\lambda_1\|\theta\|_1$):** Força sparsity (alguns θ = 0)
- **Termo L² ($\lambda_2\|\theta\|^2$):** Estabilidade e encolhimento uniforme

**Vantagens:**
- ✓ Seleção de features (como Lasso)
- ✓ Estabilidade (como Ridge)
- ✓ Funciona melhor quando features são correlacionadas

**Trade-off:**
- ✗ Dois hiperparâmetros para tunar ($\lambda_1$ e $\lambda_2$)
- ✗ Mais complexo de otimizar

## Questão 3.3: Early Stopping

**Nível:** Médio  
**Tema:** Regularização Temporal

O que é Early Stopping? Por que funciona como regularização?

**Resposta:**

**Early Stopping:**
- Durante treinamento iterativo, monitora um **validation set**
- Quando validation error começa a **aumentar**, **PARA o treinamento**
- Não espera até error de treino ir para zero

**Por que funciona:**

```
Erro
  |
  |  \ Treino (continua diminuindo)
  |   \
  |    \ | ← PARA AQUI!
  |     \|  Validação (começa a subir)
  |      /
  |_____/__________  Epochs
```

**Mecanismo:**
- Nas primeiras iterações: Aprende padrões reais
- Depois: Começa a se ajustar ao ruído (overfit)
- Early Stopping para no ponto certo antes do overfit

**Equivalência:**
- Para modelos lineares, provou-se que Early Stopping é **equivalente a L² Regularization**
- Parar cedo ≈ Ter parâmetros menores

## Questão 3.4: Cross-Validation Prático

**Nível:** Médio  
**Tema:** Seleção de Hiperparâmetros

Descreva passo a passo como usar 5-Fold CV para escolher λ em Ridge Regression.

**Resposta:**

**Procedimento:**

1. **Defina candidatos para λ:**
   - λ ∈ {0.0001, 0.001, 0.01, 0.1, 1, 10, 100}
   - Use escala logarítmica (spread exponencial)

2. **Para cada λ, faça 5-Fold CV:**
   - Divide dados em 5 folds
   - Para cada fold i:
     - Treina Ridge(λ) nos outros 4 folds
     - Valida no fold i, registra erro
   - Calcula erro médio:
     $$CV\text{-}Error(\lambda) = \frac{1}{5}\sum_{i=1}^5 \text{erro}_i(\lambda)$$

3. **Escolha λ ótimo:**
   - $$\lambda^* = \arg\min_\lambda CV\text{-}Error(\lambda)$$

4. **Retreine com λ*:**
   - Use TODOS os dados (treino + validação)
   - Treina Ridge(λ*)

5. **Teste no Test Set:**
   - Avalia modelo final no test set (nunca visto antes)

**Importante:**
- ✓ CV usa melhor os dados disponíveis
- ✓ Estimativa mais estável que hold-out simples
- ✗ Nunca use test set para escolher λ (contamina avaliação!)

## Questão 3.5: Variantes de Cross-Validation

**Nível:** Fácil  
**Tema:** Métodos CV

Descreva brevemente:
a) Hold-Out (2-Fold)
b) 5-Fold vs 10-Fold
c) Stratified K-Fold
d) Leave-One-Out (LOO)

**Resposta:**

**a) Hold-Out:**
- Divide dados em 2: 70% treino, 30% validação
- Simples mas menos estável (só 1 validação)
- Rápido

**b) K-Fold:**
- **5-Fold:** Padrão equilibrado (não muito caro, razoavelmente estável)
- **10-Fold:** Mais estável mas mais lento (10×)
- Trade-off: precisão vs custo computacional

**c) Stratified K-Fold:**
- Mantém proporções de classes em cada fold
- Importante para classificação com classes desbalanceadas
- Se 80% classe A e 20% classe B no total, cada fold tem 80-20

**d) Leave-One-Out (LOO):**
- K = número de observações (n)
- Cada observação é 1 fold de validação
- Treina n vezes (muito caro!)
- Mais estável teoricamente mas impraticável para n grande

# AULA 4: VISÃO PROBABILÍSTICA

## Questão 4.1: Teorema de Bayes

**Nível:** Médio  
**Tema:** Interpretação Probabilística

Enuncie o Teorema de Bayes e explique cada termo no contexto de regressão.

**Resposta:**

$$p(\theta | D) = \frac{p(D | \theta) \cdot p(\theta)}{p(D)}$$

**Termos:**

- **$p(\theta|D)$ - Posterior:** 
  - O que sabemos sobre θ DEPOIS de ver os dados
  - Isto é o que queremos estimar!

- **$p(D|\theta)$ - Likelihood:**
  - Probabilidade de observar esses dados dado θ
  - Função de FIT: dados bem explicados → likelihood alta
  - Em regressão Gaussiana: $p(D|\theta) \propto \exp(-\|B\theta-y\|^2)$

- **$p(\theta)$ - Prior:**
  - Conhecimento prévio sobre θ ANTES dos dados
  - Codifica nossas preferências:
    - Ridge: Prior Gaussiano (preferimos θ pequeno)
    - Lasso: Prior Laplaciano (preferimos θ = 0)

- **$p(D)$ - Evidência:**
  - Probabilidade total dos dados
  - Normalizador (integral sobre todos θ)
  - Frequentemente ignorada em estimação pontual

## Questão 4.2: MLE

**Nível:** Médio  
**Tema:** Maximum Likelihood Estimation

O que é MLE? Como se relaciona com Equação Normal para regressão?

**Resposta:**

**MLE (Maximum Likelihood Estimation):**
- Encontre θ que torna os dados observados **mais prováveis**
- $$\theta^*_{MLE} = \arg\max_\theta p(D|\theta)$$

**Para regressão com ruído Gaussiano:**

Assumindo: $y_i = B_i \theta + \epsilon_i$ com $\epsilon_i \sim N(0, \sigma^2)$

A likelihood é:
$$p(D|\theta) = \prod_i \exp\left(-\frac{(y_i - B_i\theta)^2}{2\sigma^2}\right)$$

Tomando $-\log p(D|\theta)$:
$$-\log p(D|\theta) \propto \sum_i (y_i - B_i\theta)^2 = \|B\theta-y\|^2$$

**Resultado:** 
$$\theta^*_{MLE} = (B^T B)^{-1}B^T y$$

**Conclusão:** MLE é equivalente a minimizar MSE (que é a Equação Normal)!

---

## Questão 4.3: MAP

**Nível:** Médio  
**Tema:** Maximum A Posteriori

Qual é a diferença entre MLE e MAP?

**Resposta:**

**MLE (sem prior):**
$$\theta^*_{MLE} = \arg\max_\theta p(D|\theta)$$
- Só olha likelihood (fit aos dados)
- Sem regularização implícita
- Pode overfitting com dados limitados

**MAP (com prior):**
$$\theta^*_{MAP} = \arg\max_\theta p(\theta|D) = \arg\max_\theta [p(D|\theta) \cdot p(\theta)]$$
- Combina likelihood E conhecimento prévio (prior)
- Inclui regularização via prior
- Mais robusto com dados limitados

**Relação:**
$$-\log p(\theta|D) = -\log p(D|\theta) - \log p(\theta) + const$$
$$= \text{Data Loss} + \text{Regularizer} + const$$

**Conclusão:** Regularização é exatamente MAP com escolha apropriada de prior!

## Questão 4.4: Ridge é Prior Gaussiano

**Nível:** Difícil  
**Tema:** Prior → Regularizador

Mostre como Prior Gaussiano $p(\theta) \propto \exp(-\lambda\|\theta\|^2)$ leva a Ridge Regression.

**Resposta:**

**Passo 1: Prior Gaussiano**
$$p(\theta) = C \exp(-\lambda\|\theta\|^2)$$
(C é constante de normalização)

**Passo 2: Likelihood Gaussiana**
$$p(D|\theta) \propto \exp(-\|B\theta-y\|^2)$$

**Passo 3: Posterior (Bayes)**
$$p(\theta|D) \propto p(D|\theta) \cdot p(\theta)$$
$$\propto \exp(-\|B\theta-y\|^2) \cdot \exp(-\lambda\|\theta\|^2)$$
$$= \exp(-\|B\theta-y\|^2 - \lambda\|\theta\|^2)$$

**Passo 4: MAP**
$$\theta^*_{MAP} = \arg\max_\theta \log p(\theta|D)$$
$$= \arg\max_\theta [-\|B\theta-y\|^2 - \lambda\|\theta\|^2]$$
$$= \arg\min_\theta [\|B\theta-y\|^2 + \lambda\|\theta\|^2]$$

**Isto é Ridge Regression!**

**Solução:** $(B^T B + \lambda I)^{-1}B^T y$

---

## Questão 4.5: Lasso é Prior Laplaciano

**Nível:** Difícil  
**Tema:** Prior Laplaciano → Sparsity

Explique como Prior Laplaciano leva a Lasso e por que causa sparsity.

**Resposta:**

**Prior Laplaciano:**
$$p(\theta) \propto \exp(-\lambda\|\theta\|_1)$$
onde $\|\theta\|_1 = |\theta_1| + |\theta_2| + \ldots + |\theta_p|$

**Derivação (similar ao Ridge):**
$$p(\theta|D) \propto \exp(-\|B\theta-y\|^2 - \lambda\|\theta\|_1)$$

$$\theta^*_{MAP} = \arg\min_\theta [\|B\theta-y\|^2 + \lambda\|\theta\|_1]$$

**Isto é Lasso Regression!**

**Por que causa sparsity (zeros)?**

1. **Geometricamente:**
   - Ridge: Restrição em esfera (smooth)
   - Lasso: Restrição em losango (tem vértices nos eixos!)
   - Solução ótima frequentemente cai num vértice
   - Vértice = alguns θ_j = 0

2. **Probabilisticamente:**
   - Laplace tem **pico agudo em 0** (grande densidade)
   - Muito mais "preferência" por θ = 0 comparado a Gaussiano
   - Likelihood precisa ser muito boa para justificar θ ≠ 0

**Resultado:** Seleção automática de features (feature selection)!

## Questão 5.1: Gradient Descent Básico

**Nível:** Fácil  
**Tema:** GD Fundamentals

O que é gradiente? Como é usado em Gradient Descent?

**Resposta:**

**Gradiente:**
- Vetor de derivadas parciais
- $\nabla_\theta \mathcal{L} = (\frac{\partial \mathcal{L}}{\partial \theta_0}, \frac{\partial \mathcal{L}}{\partial \theta_1}, \ldots)$
- Aponta na direção de **MAIOR AUMENTO** da loss

**Gradient Descent:**
$$\theta_{t+1} = \theta_t - \eta \nabla_\theta \mathcal{L}(\theta_t)$$

- Vamos na direção **OPOSTA** ao gradiente (descida)
- $\eta$ = learning rate (tamanho do passo)
- Repetimos até convergência

**Intuição:**
- Como descer uma montanha no escuro:
  - Sinta a inclinação (gradiente)
  - Dê um passo para baixo (direção oposta)
  - Repita

## Questão 5.2: Momentum vs Nesterov

**Nível:** Médio  
**Tema:** Otimizadores Avançados

Qual é a diferença entre Momentum e Nesterov Acceleration?

**Resposta:**

**Vanilla Momentum:**
```
v_{t+1} = β·v_t - η·∇L(θ_t)      ← calcula gradiente AQUI
θ_{t+1} = θ_t + v_{t+1}
```

**Nesterov Acceleration:**
```
v_{t+1} = β·v_t - η·∇L(θ_t + β·v_t)  ← calcula gradiente ADIANTE
θ_{t+1} = θ_t + v_{t+1}
```

**Diferença chave:**
- **Vanilla:** Calcula gradiente em θ_t, depois avança
- **Nesterov:** "Antecipa" para θ_t + β·v_t, calcula gradiente lá, depois avança

**Vantagem Nesterov:**
- ✓ Converge mais rápido
- ✓ Menos oscilações em vales alongados
- ✓ Melhor desempenho teórico

**Analogia:**
- Vanilla: Você vê a encosta AQUI, depois se move
- Nesterov: Você "antecipa" a encosta ONDE vai estar, depois se move

## Questão 5.3: Mini-Batch SGD

**Nível:** Fácil  
**Tema:** Batch Processing

Por que usar mini-batches em vez de todo o dataset a cada iteração?

**Resposta:**

**Razões Computacionais:**

Com N = 1.000.000 amostras:
- Batch completo: Calcular gradiente em 1M dados = lento
- Mini-batch de 32: Calcular em 32 dados = 31.250× mais rápido!
- Múltiplos passos rápidos > poucos passos lentos

**Razões Teóricas:**

- ✓ Ruído do sampling ajuda a **escapar mínimos locais**
- ✓ Redes neurais têm muitos mínimos locais ruins
- ✓ SGD barulhento consegue "pular" para regiões melhores

**Trade-off:**
- ✓ Muito mais rápido
- ✓ Cabe em GPU/memória
- ✗ Convergência menos estável (ruidosa)
- ✗ Precisa ajuste cuidadoso de learning rate

**Tamanhos recomendados:**
- B = 32 ou 64: Padrão para deep learning
- B = 128: Para datasets muito grandes
- B = N: Batch completo (rare hoje em dia)

**R:** 
$$f_\theta(x) = \sum_{j=0}^{p} \theta_j B_j(x)$$

Matriz de design:
$$\mathbf{B} = \begin{pmatrix} B_0(x_1) & B_1(x_1) & \cdots & B_p(x_1) \\ \vdots & \vdots & \ddots & \vdots \\ B_0(x_n) & B_1(x_n) & \cdots & B_p(x_n) \end{pmatrix} \in \mathbb{R}^{n \times (p+1)}$$

Vetor de parâmetros: $\boldsymbol{\theta} = (\theta_0, \theta_1, \ldots, \theta_p)^T \in \mathbb{R}^{p+1}$

---

### (b) (2 pontos)
**Q:** Explique o que significa dizer que o modelo é *linear nos parâmetros* mas possivelmente *não-linear em $x$*. Por que essa propriedade é importante?

**R:**

**Linear nos parâmetros:** Os parâmetros $\theta_j$ aparecem linearmente na expressão de $f_\theta(x)$. Isto é, $f$ é combinação linear dos $\theta_j$.

**Não-linear em $x$:** O argumento $x$ pode aparecer de forma não-linear dentro das funções de base $B_j(x)$.

**Exemplos:**
- $f(x) = \theta_0 + \theta_1 x^2 + \theta_2 \sin(x)$ : Linear em $\theta$, não-linear em $x$
- $f(x) = \theta_0 + \theta_1 x$ : Linear em ambos
- $f(x) = \theta_0 e^{\theta_1 x}$ : Não-linear em $\theta$

**Importância:** Se o modelo é linear nos parâmetros, podemos usar **álgebra linear** (Equação Normal) para encontrar a solução ótima de forma **fechada e eficiente**, sem iterações.

### (b) (3 pontos)
**Q:** Explique a regularização L² (Tikhonov) em detalhes. O que é o parâmetro $\lambda$? Como afeta a solução?

**R:** [Resposta resumida de 3.1]

$$\mathcal{L}(\theta) = \frac{1}{N}\|B\theta - y\|^2 + \lambda \|\theta\|^2$$

- $\lambda = 0$: Sem regularização
- $\lambda$ grande: Forte regularização (θ ≈ 0)
- Solução: $\theta_\lambda = (B^T B + \lambda I)^{-1}B^T y$
- Efeito: Encolhe TODOS parâmetros, torna B^T B invertível
